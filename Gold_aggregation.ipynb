# GOLD LAYER 


```python
from pyspark.sql.functions import *
# Define Active Storage Path
active_path = "abfss://active-data@strgarchiveproject01.dfs.core.windows.net/"
```

```python
# READ SILVER TABLE
silver_df = spark.read.format("delta") \
    .load(active_path + "silver_ecommerce")
```

```python
# BUSINESS AGGREGATIONS
gold_df = silver_df.groupBy(
    "CustomerID",
    "Country"
).agg(
    sum("Quantity").alias("TotalQuantity"),
    round(sum("UnitPrice"), 2).alias("TotalAmount")
)
# Preview data
display(gold_df)
```

```python
# SAVE GOLD TABLE
gold_df.write.format("delta") \
    .mode("overwrite") \
    .save(active_path + "gold_customer_summary")

print("Gold Layer Created Successfully")
```
