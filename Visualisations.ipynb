```python
#VISUALIZATION 1 — Bronze vs Silver vs Archive Record Count
bronze_count = spark.read.format("delta").load(
    "abfss://active-data@strgarchiveproject01.dfs.core.windows.net/bronze_ecommerce"
).count()

silver_count = spark.read.format("delta").load(
    "abfss://active-data@strgarchiveproject01.dfs.core.windows.net/silver_ecommerce"
).count()

archive_count = spark.read.format("delta").load(
    "abfss://archive-data@strgarchiveproject01.dfs.core.windows.net/archived_silver"
).count()

summary_df = spark.createDataFrame([
    ("Bronze", bronze_count),
    ("Silver", silver_count),
    ("Archive", archive_count)
], ["Layer", "Count"])

display(summary_df)
```

```python
#VISUALIZATION 2 — Archived Records by Country
from pyspark.sql.functions import count

archive_df = spark.read.format("delta").load(
    "abfss://archive-data@strgarchiveproject01.dfs.core.windows.net/archived_silver"
)

country_df = archive_df.groupBy("Country") \
    .agg(count("*").alias("ArchivedRecords")) \
    .orderBy("ArchivedRecords", ascending=False)

display(country_df)
```

```python
#VISUALIZATION 3 — Monthly Archived Records Trend
from pyspark.sql.functions import year, month

archive_df = spark.read.format("delta").load(
    "abfss://archive-data@strgarchiveproject01.dfs.core.windows.net/archived_silver"
)

trend_df = archive_df.groupBy(
    year("InvoiceDate").alias("Year"),
    month("InvoiceDate").alias("Month")
).count().orderBy("Year", "Month")

display(trend_df)
```
